# Suite CONF — Conformance across implementations

Every implementation writes the same files for the conformance corpus (`conformance/`): the certificates of one
fixture's run, suite RRN's (a session over a shelf of `bolt` and `nut`, each candidate of `Label` applied with the
caller's choice and certified), as mbse-schemas' JSON snapshot of the certificates' store
(`tests/conformance.py`). This suite checks that this implementation's files are current, and that they are
byte-identical to every other implementation's: the certificates are the same data on the wire in either language.

```python
from conformance import ROOT, render

MINE = ROOT / "python3"
OTHERS = sorted(d for d in ROOT.iterdir() if d.is_dir() and d.name != "python3")
files = render()
```

<a id="CONF-01"></a>
## CONF-01 · This implementation's files are current

```python
assert sorted(p.name for p in MINE.iterdir()) == sorted(files)
for name, text in files.items():
    assert (MINE / name).read_text(encoding="utf-8") == text, f"{name} is stale: run uv run python tests/conformance.py"
print(sorted(files))
```

<a id="CONF-02"></a>
## CONF-02 · Every other implementation wrote the same bytes

```python
assert [d.name for d in OTHERS] == ["typescript5"]
for other in OTHERS:
    assert sorted(p.name for p in other.iterdir()) == sorted(files)
    for name, text in files.items():
        assert (other / name).read_text(encoding="utf-8") == text, f"{other.name}/{name} differs"
print("identical to", [d.name for d in OTHERS])
```
